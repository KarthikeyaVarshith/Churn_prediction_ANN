```python
import pandas as pd
import numpy as np
import pickle
from tensorflow.keras.models import load_model

```

```python
model =  load_model('model.h5')
```

```python
with open ('scaler.pkl', 'rb') as file:
    scaler = pickle.load(file)

with open ('ohe.pkl', 'rb') as file:
    ohe = pickle.load(file)

with open ('lencoder.pkl', 'rb') as file:
    le = pickle.load(file)
```

```python
ipdata = {
    "CreditScore" : 600,
    "Geography": 'France',
    "Gender": "Male",
    "Age": 40,
    "Tenure": 3,
    "Balance": 60000,
    "NumOfProducts": 2,
    "HasCrCard": 1,
    "IsActiveMember": 1,
    "EstimatedSalary": 50000,

}
```

```python
ipdata_n = {
    "CreditScore" :619,
    "Geography": 'France',
    "Gender": "Female",
    "Age": 42,
    "Tenure": 2,
    "Balance": 0.0,
    "NumOfProducts": 1,
    "HasCrCard": 1,
    "IsActiveMember": 1,
    "EstimatedSalary": 101348.88,

}
```

```python
ip_df = pd.DataFrame([ipdata])
ip_df
```

```python
ip_df['Gender'] = le.transform(ip_df['Gender'])
```

```python
geo_en_df = pd.DataFrame(ohe.transform(ip_df[['Geography']]).toarray(),columns=ohe.get_feature_names_out())
```

```python
ip_df = pd.concat([ip_df.drop(columns="Geography"), geo_en_df], axis=1)
ip_df
```

```python
ip_df.columns
```

```python
ip_df = ip_df[['CreditScore', 'Gender', 'Age','Geography_France', 'Geography_Germany', 'Geography_Spain', 'Tenure', 'Balance', 'NumOfProducts',
       'HasCrCard', 'IsActiveMember', 'EstimatedSalary']]
```

```python
ip_scaled = scaler.transform(ip_df)
ip_scaled
```

```python
model.predict(ip_scaled)[0][0]
```
